```python
import sys, platform
print(sys.version)
print(platform.platform())
```

```python
import importlib
for name in ['numpy','pandas','sklearn','matplotlib','torch','transformers','underthesea','lime','shap']:
    try:
        m = importlib.import_module(name)
        print('OK   ', name, getattr(m, '__version__', ''))
    except Exception as e:
        print('THIẾU', name, type(e).__name__)
```

```python
try:
    import torch
    print('GPU:', torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'không có, dùng CPU')
except Exception as e:
    print('Chưa cài torch:', e)
```
